
Projet Machine Learning : La loi d’Okun

1. Introduction

La loi d’Okun étudie la relation entre la croissance économique et le taux de chômage. Ce projet vise à analyser cette relation à l’aide du Machine Learning, en utilisant un modèle de régression linéaire simple.




```python
# 2. Importation des bibliothèques

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

print("Bibliothèques importées avec succès !")
```

```python
# 3. Création du jeu de données

data = {
    'Croissance_economique': [-2, 0, 1, 2, 3, 4, 5, 6, 2.5, 3.5],
    'Taux_chomage': [12, 11, 10.5, 10, 9, 8, 7, 6, 9.5, 8.5]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 4. Visualisation des données

plt.figure(figsize=(8, 5))

plt.scatter(df['Croissance_economique'],
            df['Taux_chomage'],
            color='blue',
            label='Données')

plt.xlabel('Croissance économique (%)')
plt.ylabel('Taux de chômage (%)')
plt.title("Relation entre croissance économique et chômage")

plt.legend()
plt.grid(True)
plt.show()
```

```python
# 5. Entraînement du modèle

X = df[['Croissance_economique']]
y = df['Taux_chomage']

model = LinearRegression()

model.fit(X, y)

print("Coefficient :", model.coef_[0])
print("Intercept :", model.intercept_)
print("R² :", model.score(X, y))
```

```python
# 6. Droite de régression

y_pred = model.predict(X)

plt.figure(figsize=(8, 5))

plt.scatter(X, y, color='blue', label='Données réelles')

plt.plot(X, y_pred, color='red', label='Droite de régression')

plt.xlabel('Croissance économique (%)')
plt.ylabel('Taux de chômage (%)')

plt.title("Loi d'Okun : Régression linéaire")

plt.legend()
plt.grid(True)
plt.show()
```

```python
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)

print("MSE :", mse)
print("RMSE :", rmse)
print("R² :", r2)
```

```python
# 8. Prédiction du taux de chômage

nouvelles_croissances = pd.DataFrame({
    'Croissance_economique': [7, 8, 9, 10]
})

predictions = model.predict(nouvelles_croissances)

for croissance, chomage in zip([7, 8, 9, 10], predictions):
    print(f"Croissance {croissance}% → Chômage prédit : {chomage:.2f}%")
```

Conclusion
Ce projet a permis d'étudier la relation entre la croissance économique et le taux de chômage à travers la loi d'Okun. Nous avons utilisé un modèle de régression linéaire pour analyser cette relation. Les résultats montrent une relation négative entre les deux variables. Le modèle obtenu présente un coefficient de détermination de 96,45 % sur notre jeu de données illustratif.
Important : nos données sont simulées pour le projet ; elles ne constituent pas des observations économiques réelles.
